# Q25 Pattern Review Performance

Grouped comparison of Q25 performance with and without the pattern review method across planning domains.

```python
# Cell 1: Imports and paper-ready plotting constants
import matplotlib.pyplot as plt

TITLE_FONTSIZE = 24
AXIS_LABEL_FONTSIZE = 20
TICK_LABEL_FONTSIZE = 17
LEGEND_FONTSIZE = 18
ANNOTATION_FONTSIZE = 15
```

```python
# Cell 2: Q25 performance data (percent)
domains = ["BlocksWorld", "Elevators", "Transport", "ChildSnack"]
q25_scores = [40, 6, 66, 19]
q25_pattern_review_scores = [46, 13, 79, 51]

for domain, q25, reviewed in zip(domains, q25_scores, q25_pattern_review_scores):
    print(f"{domain:<12} Q25: {q25:>2}%  Q25 Pattern Review: {reviewed:>2}%")
```

```python
# Cell 3: Plot grouped Q25 performance bars by domain
x_positions = list(range(len(domains)))
bar_width = 0.34
q25_positions = [x - bar_width / 2 for x in x_positions]
review_positions = [x + bar_width / 2 for x in x_positions]

fig, ax = plt.subplots(figsize=(12.5, 7.5))
fig.patch.set_facecolor("#F6F1E8")
ax.set_facecolor("#FFFDF8")

q25_bars = ax.bar(
    q25_positions,
    q25_scores,
    width=bar_width,
    label="Q25",
    color="#0B7285",
    edgecolor="#075866",
    linewidth=1.2,
)
review_bars = ax.bar(
    review_positions,
    q25_pattern_review_scores,
    width=bar_width,
    label="Q25 Pattern Review",
    color="#D97706",
    edgecolor="#A95804",
    linewidth=1.2,
)

for bars, values, color in (
    (q25_bars, q25_scores, "#075866"),
    (review_bars, q25_pattern_review_scores, "#A95804"),
):
    for bar, value in zip(bars, values):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            value + 1.5,
            f"{value}%",
            ha="center",
            va="bottom",
            fontsize=ANNOTATION_FONTSIZE,
            fontweight="bold",
            color=color,
        )

ax.set_xticks(x_positions, domains)
ax.set_ylim(0, 100)
ax.set_yticks(range(0, 101, 20))
ax.set_xlabel("Domain", fontsize=AXIS_LABEL_FONTSIZE, fontweight="semibold", labelpad=12)
ax.set_ylabel("Performance (%)", fontsize=AXIS_LABEL_FONTSIZE, fontweight="semibold", labelpad=12)
ax.set_title(
    "Q25 Performance With and Without Pattern Review",
    fontsize=TITLE_FONTSIZE,
    fontweight="bold",
    pad=18,
)
ax.grid(axis="y", color="#D8D1C4", linewidth=0.8, alpha=0.8)
ax.set_axisbelow(True)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="both", labelsize=TICK_LABEL_FONTSIZE)
ax.legend(frameon=False, fontsize=LEGEND_FONTSIZE, loc="upper left")
fig.tight_layout()
plt.show()
```
