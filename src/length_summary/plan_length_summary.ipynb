# Plot Block vs. Plan relation

```python
# Cell 1: Setup imports, paths, and configurable handles
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Project root (run notebook from formalizer-scaling/src/length_summary)
PROJECT_ROOT = Path.cwd().parents[1]
OUTPUT_ROOT = PROJECT_ROOT / "output"

# Flexible handles
PIPELINE_NAME = "direct_pddl"                     # e.g., direct_pddl, direct_pddl_unraveling
MODEL_NAME = "qwen3_32b_openrouter_retry3"                     # e.g., gemini_3_flash, qwen2_5_32b_instruct_kani_retry3
BLOCK_COUNTS = [5, 30, 50, 75]                    # e.g., [5, 30, 50, 75, 100]

# Fixed mapping from block-count batch to problem-id range (10 problems each)
BLOCK_TO_PROBLEM_RANGE = {
  5: (1, 10),
  10: (11, 20),
  15: (21, 30),
  20: (31, 40),
  25: (41, 50),
  30: (51, 60),
  35: (61, 70),
  40: (71, 80),
  45: (81, 90),
  50: (91, 100),
  55: (101, 110),
  60: (111, 120),
  65: (121, 130),
  70: (131, 140),
  75: (141, 150),
  80: (151, 160),
  85: (161, 170),
  90: (171, 180),
  95: (181, 190),
  100: (191, 200),
}

MODEL_DIR = OUTPUT_ROOT / PIPELINE_NAME / MODEL_NAME

print("PROJECT_ROOT:", PROJECT_ROOT)
print("MODEL_DIR:", MODEL_DIR)
print("MODEL_DIR exists:", MODEL_DIR.exists())
print("BLOCK_COUNTS:", BLOCK_COUNTS)
```

```python
# Cell 2: Build target problem IDs from selected block counts
def pid_from_int(i: int) -> str:
  return f"p{i:02d}" if i < 100 else f"p{i}"

target_rows = []
for block_count in BLOCK_COUNTS:
  if block_count not in BLOCK_TO_PROBLEM_RANGE:
      raise ValueError(f"Unsupported block count: {block_count}. Add it to BLOCK_TO_PROBLEM_RANGE.")
  start_i, end_i = BLOCK_TO_PROBLEM_RANGE[block_count]
  for i in range(start_i, end_i + 1):
      target_rows.append(
          {
              "problem_id": pid_from_int(i),
              "problem_int": i,
              "block_count": block_count,
          }
      )

TARGET_DF = pd.DataFrame(target_rows).sort_values(["block_count", "problem_int"]).reset_index(drop=True)

print("Total target problems:", len(TARGET_DF))
print(TARGET_DF.head(12).to_string(index=False))
```

```python
# Cell 3: Collect plan line counts (only problems with plan.txt)
records = []
for _, row in TARGET_DF.iterrows():
  pid = row["problem_id"]
  block_count = int(row["block_count"])

  plan_path = MODEL_DIR / pid / "cand_00" / "plan.txt"
  if not plan_path.exists():
      continue

  plan_text = plan_path.read_text().strip()
  if not plan_text:
      continue

  line_count = len([ln for ln in plan_text.splitlines() if ln.strip() != ""])
  records.append(
      {
          "problem_id": pid,
          "block_count": block_count,
          "plan_line_count": line_count,
          "plan_path": str(plan_path),
      }
  )

PLAN_DF = pd.DataFrame(records).sort_values(["block_count", "problem_id"]).reset_index(drop=True)

print("Problems with plan.txt:", len(PLAN_DF))
if len(PLAN_DF):
  print(PLAN_DF.head(20).to_string(index=False))
else:
  print("No plan.txt found for selected pipeline/model/block counts.")
```

```python
# Cell 4: Compute average plan length per block count
if len(PLAN_DF) == 0:
  raise RuntimeError("PLAN_DF is empty. Run Cell 3 and verify selected pipeline/model/block counts.")

AVG_DF = (
  PLAN_DF.groupby("block_count", as_index=False)["plan_line_count"]
  .mean()
  .rename(columns={"plan_line_count": "avg_plan_line_count"})
  .sort_values("block_count")
  .reset_index(drop=True)
)

print("Average plan length by block count:")
print(AVG_DF.to_string(index=False))
```

```python
# Cell 5: Line plot of average plan length (annotated with integer y-values)
if "AVG_DF" not in globals() or len(AVG_DF) == 0:
  raise RuntimeError("AVG_DF is empty. Run Cell 4 first.")

plot_df = AVG_DF.copy()
plot_df["avg_plan_line_count_int"] = plot_df["avg_plan_line_count"].round().astype(int)

plt.figure(figsize=(9, 6))
plt.plot(
  plot_df["block_count"],
  plot_df["avg_plan_line_count_int"],
  marker="o",
  linewidth=2,
)

for _, row in plot_df.iterrows():
  x = int(row["block_count"])
  y = int(row["avg_plan_line_count_int"])
  plt.annotate(
      str(y),
      (x, y),
      textcoords="offset points",
      xytext=(0, 8),   # 8 points above
      ha="center",
      va="bottom",
  )

plt.title(f"Average Plan Length | {PIPELINE_NAME} / {MODEL_NAME}")
plt.xlabel("Number of Blocks")
plt.ylabel("Average Number of Lines in plan.txt")
plt.grid(True, linestyle="--", alpha=0.35)

plt.show()
```
